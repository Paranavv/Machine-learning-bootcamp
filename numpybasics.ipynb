```python
n = 1_000_000

pyList = list(range(n))
numpyArr = np.arange(n)

print(len(pyList), len(numpyArr))
print(pyList[:5], numpyArr[:5])
```

```python
t1 = time.time()
pyResult = [x * 2 for x in pyList]
t2 = time.time()

t3 = time.time()
numpyResult = numpyArr * 2
t4 = time.time()

print(t2 - t1, t4 - t3)
```
